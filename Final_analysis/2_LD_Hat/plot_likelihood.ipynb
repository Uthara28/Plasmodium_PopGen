#### BinhPhuoc chromosome 14

```python
filename = '/data/proj2/home/students/u.srinivasan/Plasmodium_us/Inputs/SMC/LD_hat_rho_mu_data/w_missingness/bp_chrom14/outfile.txt'
df = pd.read_csv(filename, sep='\s+', skiprows=10, header=None, usecols=[0, 1],names=['4Ner(region)', 'Pairwise_Lk'])
df = df.iloc[1:100,]
# Read data from a text file (tab or space separated)
# Adjust sep='\t' or sep=' ' based on your file format

# Inspect the first few rows to understand the data
print(df.head())

# Plot: first column (x) vs second column (y)
plt.figure(figsize=(8, 5))
plt.plot(df.iloc[:, 0], df.iloc[:, 1], marker='o', linestyle='-')

# Set x-axis ticks every 25 units
plt.xticks(range(0, int(df.iloc[:, 0].max()) + 1, 25))

# Set y-axis ticks every 0.002 units using numpy
y_min = df.iloc[:, 1].min()
y_max = df.iloc[:, 1].max()
plt.yticks(np.arange(y_min, y_max + 0.001, 0.02))

plt.xlabel(df.columns[0])
plt.ylabel(df.columns[1])
plt.title('BinhPhuoc-Vietnam, Chromosome 14')
plt.grid(True)
plt.tight_layout()
plt.show()
```
